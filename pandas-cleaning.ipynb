## check and duplicate value

```python
print("Remove customers duplicate value")
print(df_customer.duplicated())
print(df_customer.drop_duplicates(inplace=True))
```

```python
print("Remove Orders Duplicates Value")
print(df_order.duplicated())
print(df_order.drop_duplicates(inplace=True))
```

### Check Data Type

```python
print("Check Data Type of Customer Raw Data")
print(df_customer.dtypes)
print("Check Data Type of Order Raw Date")
print(df_order.dtypes)
```

```python
# convert sign_date into date format
df_customer["signup_date"] = pd.to_datetime(
        df_customer["signup_date"], 
        errors= "coerce")
```

```python
# conert order_date into date format
df_order["order_date"] = pd.to_datetime(
    df_order["order_date"],
    errors= "coerce"
)
```

```python
print(df_customer["signup_date"].dtype)
print(df_order["order_date"].dtype)
```

### check missing value

```python
print(df_customer.isnull().sum())
```

```python
print(df_order.isnull().sum())
```

```python
# filling customer_id null value with Unknown
df_order["customer_id"] = df_order["customer_id"].fillna("Unknown")
```

```python
# fill product_id null value replace with Unknown
df_order["product_id"] = df_order["product_id"].fillna("Unknown")
```

```python
print(df_order.isnull().sum())
```

### Handle order date missing date

```python
df_order[df_order["order_date"].isnull()]
```

```python
df_order["order_date_status"] = df_order["order_date"].isna().map({
    True: "Missing",
    False: "Valid"
})
```

```python
print(df_order[["order_id", "order_date", "order_date_status"]].head())
```

## cleaned file into SQL

```python
from sqlalchemy import create_engine
engine = create_engine(
    "mysql+pymysql://root:atif@localhost/sales"
)
```
